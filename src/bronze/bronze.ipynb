## Read data from the csv files and write bronze tables

```python
# read and write data by loop through the config
for item in ingestion_config:
    # reading files
    print(f"Reading from {item['path']}")
    df = (
        spark.read
            .option("header", "true")
            .option("inferSchema", "true")
            .csv(item["path"])
    )
               
    # writing to table
    print(f"Writing to {item['table']}")
    (  
     df.write
        .mode("overwrite")
        .format("delta")
        .saveAsTable(f"workspace.bronze.{item['table']}")
    )
```
